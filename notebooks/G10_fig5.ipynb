# G10_fig5: the most confident mistakes on the test set

Makes figure 5 of the report with `report/make_report_figs.py`. Nothing is trained and no GPU is needed.

It needs `KAGGLE_USERNAME` and `KAGGLE_KEY` in the Colab secrets, to download the images.

## 1. Get the code

```python
import os, subprocess
from pathlib import Path

REPO = Path("/content/dlvr-g10")
BRANCH = "audit-fixes"   # change to "main" when the branch is merged
if not REPO.exists():
    subprocess.run(["git", "clone", "--quiet", "--branch", BRANCH,
                    "https://github.com/Mikkelflassen/dlvr-g10.git", str(REPO)], check=True)
os.chdir(REPO)
subprocess.run(["pip", "install", "-q", "torchmetrics", "kaggle"], check=True)
```

## 2. Get the images

ISIC 2020 (Rotemberg et al., 2021), resized by C. Deotte: https://www.kaggle.com/datasets/cdeotte/jpeg-melanoma-256x256

```python
# skips the download if the images are already there
from google.colab import userdata
os.environ["KAGGLE_USERNAME"] = userdata.get("KAGGLE_USERNAME")
os.environ["KAGGLE_KEY"] = userdata.get("KAGGLE_KEY")

DATA = Path("/content/isic")
if not (DATA / "train").exists():
    subprocess.run(["kaggle", "datasets", "download", "-d", "cdeotte/jpeg-melanoma-256x256",
                    "-p", "/content", "-q"], check=True)
    subprocess.run(["unzip", "-q", "/content/jpeg-melanoma-256x256.zip", "-d", str(DATA)], check=True)
print(len(list((DATA / "train").glob("*.jpg"))), "jpgs")
```

## 3. Make the figures

The script makes figures 1-5. Figure 5 is `fig5_test_mistakes`.

```python
OUT = Path("/content/figures")
subprocess.run(["python", "report/make_report_figs.py", ".", str(OUT), str(DATA / "train")], check=True)

import pandas as pd
from IPython.display import Image, display
display(Image(str(OUT / "fig5_test_mistakes.png")))
pd.read_csv(OUT / "fig5_test_mistakes.csv")
```
